```python
import numpy as np
from scipy.interpolate import interp1d
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 10)
y = x**2 * np.sin(x)
plt.scatter(x, y)
# plt.xlabel('x')
# plt.ylabel('y')
# plt.title('x vs y')
# plt.show()
```

```python
f = interp1d(x, y, kind = 'linear')
x_dense = np.linspace(0, 10, 200)
y_dense = f(x_dense)
plt.plot(x_dense, y_dense)
```

```python

f = interp1d(x, y, kind = 'cubic')
x_dense = np.linspace(0, 10, 200)
y_dense = f(x_dense)
plt.plot(x_dense, y_dense)
```

```python
from scipy.optimize import curve_fit
func = lambda x,a,b : a*(x**2) + b
popt, pcov = curve_fit(func, x, y, p0 = (1,1))
# Plot fitted curve
plt.plot(x, func(x, *popt), color="red", label="Fit")
```

```python
import numpy as np
from scipy import special
import mpmath as mp

x = np.float64(713.0)

mp.mp.dps = 50
expected = float(mp.besseli(0, mp.mpf("713.0")))
actual = special.i0(x)

print("input:", x)
print("SciPy result:", actual)
print("mpmath reference:", expected)
print("float64 maximum:", np.finfo(np.float64).max)

assert np.isfinite(expected)
assert np.isfinite(actual)
assert np.isclose(actual, expected, rtol=1e-12)
```
