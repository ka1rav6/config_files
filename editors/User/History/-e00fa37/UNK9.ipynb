```python

import pandas as pd
from datetime import datetime
INTERVAL = 10
FILE:str = "schedule.xlsx"

df = pd.read_excel("schedule.xlsx", sheet_name="ScheduleForAnalysis", skiprows = 1)
print(df.head())
print(type(df["SDTD"][0]))
```

```python
for i in range(len(df)):
    df["SDTD"][i] = df["SDTD"][i].to_pydatetime()
    

    
    
```

```python
def getRollingSum():
    resultingSum = []
    for i in range(len(df["SDTD"])):
        if i < INTERVAL -1:
            resultingSum.append(0)
        else:
            summ = df["SDTD"][i - INTERVAL + 1: i + 1]
            resultingSum.append(sum(summ))
    return resultingSum
```

```python
import matplotlib.pyplot as pt
rollSum = getRollingSum()
pt.plot(rollSum)
pt.show()
```
